```python
from ollama import chat
# Regular response
response = chat(model = 'qwen3:4b', messages=[{'role': 'user', 'content': 'Why is the sky blue?'}], stream=True)

for i in response:
    print(i['message']['content'], end = '')
```

```python
messages = [
    {
        'role' : 'user',
        'content' : 'I want you to act like you are Jarvis assistant of Tony Stark, which means you should be good at questions related to tech and economy.',
    },
]

# for i in chat(model='qwen2.5:3b', messages=messages, stream = True):
#     print(i['message']['content'], end='')

while True:
    user_inp = input("Type your message: ")
    if (user_inp.lower() in ['bye', 'exit', 'quit', 'out', 'stop']):
        break
    response = chat(model='qwen2.5:3b', 
                    messages=[*messages, 
                              {'role': 'user',
                               'content': user_inp 
                               }
                    ],
                    stream = True)

    # Add the response to the messages to maintain the history
    print(f"User: {user_inp}\n")
    print("Assistant's response: \n")
    
    resp = ""
    for i in response:
        resp += i['message']['content']
        print(i['message']['content'], end='')
    print('\n')

    messages += [
        {'role': 'user', 'content': user_inp},
        {'role': 'assistant', 'content': resp},
    ]

print("*** Chat terminated successfully. ***")
    
```

```python
# Initialize recognizer class (for recognizing the speech)
r = sr.Recognizer()
r.pause_threshold = 1
```

```python
# For testing with live recording
with sr.Microphone() as source:
    print("Talk")
    audio_text = r.listen(source, phrase_time_limit=10)
    print("Time over, thanks")
    # recoginze_() method will throw a request
    # error if the API is unreachable,
    # hence using exception handling
    
    try:
        # using google speech recognition
        print("Input recognized from recording: "+r.recognize_google(audio_text))
    except:
        print("Sorry, I did not get that")

response = chat(model = 'qwen2.5:3b', messages=[{'role': 'user', 'content': r.recognize_google(audio_text)}], stream=True)

for i in response:
    print(i['message']['content'], end = '')
```
