```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft
from scipy.signal import spectrogram
import soundfile as sf
```

```python
file = "for_whisper.wav"

form, rate = librosa.load(file, sr = None)

print(f"form is {form}, rate is {rate}")
```

```python
from IPython.display import Audio

Audio(form, rate=rate)
```

```python
from IPython.display import Audio

Audio(form, rate=rate*2)
```

```python
from IPython.display import Audio

Audio(form, rate=rate/2)
```

```python
# Compute the amplitude range
amplitude_range = np.max(form) - np.min(form)
# Get the bit depth from the audio data
audio_data, sampling_rate = sf.read(file)
bit_depth = audio_data.dtype.itemsize
# printing values
print(f'Amplitude Range: {amplitude_range}')
print(f'Bit Depth: {bit_depth} bits')
```

```python
plt.figure(figsize=(8,4))

librosa.display.waveshow(form, sr=rate)
plt.ylabel('Amplitude')
plt.title('Audio Waveform')
plt.grid(True)
plt.show()
```
